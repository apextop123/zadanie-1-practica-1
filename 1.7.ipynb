```python
n = int(input("Введите основание: "))
m = int(input("Введите степень: "))
print(n ** m)
```

**Алгоритм** — это идея возведения числа в степень: получить два числа, возвести первое во второе, вывести результат.

**Программа** — это конкретная реализация этого алгоритма на языке Python:
n = int(input())
m = int(input())
print(n ** m)
