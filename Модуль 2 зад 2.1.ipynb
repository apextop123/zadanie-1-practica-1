```python
# 1. int — целое число
my_int = 42

# 2. float — число с плавающей точкой
my_float = 3.14

# 3. str — строка
my_str = "строка"

# 4. bool — логический тип
my_bool = True

# 5. list — список (изменяемый, упорядоченный)
my_list = [1, 2, 3]

# 6. tuple — кортеж (неизменяемый, упорядоченный)
my_tuple = (1, 2, 3)

# 7. dict — словарь (пары ключ: значение)
my_dict = {"имя": 1, "груп": 2}

# 8. set — множество (уникальные, неупорядоченные элементы)
my_set = {1, 2, 3}
```
